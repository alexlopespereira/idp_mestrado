### 6.5 Considerando o dataframe abaixo, adicione uma coluna a esse dataframe para guardar o valor do código IBGE que está junto com o nome do município na coluna Município.
#### Dica: use a função apply

```python
def adiciona_cod_ibge6(df):
    """retorne um dataframe com o código ibge de 6 dígitos numa coluna separada.
    """
    df['cod_ibge6'] = df['Município'].apply(lambda x: x.split(' ')[0])
    return df
```

```python
## Faça seus testes aqui
test_data = {'Município': {0: "110001 Alta Floresta D'Oeste",
  1: '110037 Alto Alegre dos Parecis', 2: '110040 Alto Paraíso',
  3: "110034 Alvorada D'Oeste", 4: '110002 Ariquemes'},
 '1991': {0: 0.5983, 1: 0, 2: 0, 3: 0.569, 4: 0.5827},
 '2000': {0: 0.5868, 1: 0.508, 2: 0.6256, 3: 0.6534, 4: 0.5927},
 '2010': {0: 0.5893, 1: 0.5491, 2: 0.5417, 3: 0.5355, 4: 0.5496}}
df = pd.DataFrame(test_data)
print(adiciona_cod_ibge6(df))
```

### 6.6 Escreva uma função para carregar corretamente o dataframe de morbidade disponibilizado.
#### Dica 1: No dataset de morbidade existem valores NA representados por caracteres específicos. Especifique no argumento na_values.
#### Dica 2: Especifique também o separador, o caracter de separação de casas decimais e o tipo da coluna cod_ibge6 como str.

```python
def load_morbidade(path_morbidade):
    """retorne o dataframe de morbidade.
    """
    df_morbidade = pd.read_csv(path_morbidade, sep=';', decimal=',', dtype={'cod_ibge6': str},
                               na_values=['-', '...'])
    return df_morbidade
```

```python
## Faça seus testes aqui
path_morbidade = 'https://github.com/alexlopespereira/curso_ciencia_dados2021/raw/master/data/originais/morbidade/consolidado/morbidade.csv'
print(load_morbidade(path_morbidade).head())
```

```python
# Validação
path_morbidade = 'https://github.com/alexlopespereira/curso_ciencia_dados2021/raw/master/data/originais/morbidade/consolidado/morbidade.csv'
result = load_morbidade(path_morbidade).iloc[0]
assert result['cod_ibge6'] == '110001'
print("Exercício 6.6 OK!")
```

### 6.7 Escreva uma função para carregar corretamente o dataframe de PIB de 2017 disponibilizado. Renomeie as colunas apropriadamente.
#### Dica 1: Especifique o dtype da coluna que contém o codigo do municipio para str.
#### Dica 2: Use a função rename e um dicionário para renomear as colunas.

```python
def load_pib(path_pib):
    """retorne o dataframe do pib
    """
    df_pib = pd.read_excel(path_pib, skiprows=3, skipfooter=1, dtype={'Unnamed: 1': str})
    df_pib.rename(columns={'Unnamed: 0': 'nivel', 'Unnamed: 1': 'cod_ibge7',
                           'Unnamed: 2': 'municipio'}, inplace=True)
    return df_pib
```

```python
## Faça seus testes aqui
path_pib = 'https://github.com/alexlopespereira/enapespcd2021/raw/main/data/originais/pib/pib_municipios.xlsx'
print(load_pib(path_pib).head())
```

```python
# Validação
path_pib = 'https://github.com/alexlopespereira/enapespcd2021/raw/main/data/originais/pib/pib_municipios.xlsx'
result = load_pib(path_pib).iloc[0]
assert result['nivel'] == 'MU'
assert result['cod_ibge7'] == '1100015'
print("Exercício 6.7 OK!")
```
