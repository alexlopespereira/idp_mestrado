### 2.9 Somar todos os elementos de uma lista usando for

```python
def soma_lista(lista):
    """Retorne a soma de todos os elementos da lista usando um loop for"""
    total = 0
    for elemento in lista:
        total = total + elemento
    return total
```

```python
## Faça seus testes aqui
print(soma_lista([1, 2, 3, 4]))  # 10
print(soma_lista([10, -5, 5]))   # 10
print(soma_lista([]))            # 0
```

```python
# Validação
assert soma_lista([1, 2, 3, 4]) == 10
assert soma_lista([10, -5, 5]) == 10
assert soma_lista([]) == 0
print("Exercício 2.9 OK!")
```

### 2.10 Contar quantos elementos há numa lista usando for
**Não use a função len**

```python
def contar_elementos(lista):
    """Retorne a quantidade de elementos usando um loop for (não use len)"""
    contador = 0
    for elemento in lista:
        contador = contador + 1
    return contador
```

### 2.11 Contar quantos números positivos há numa lista

```python
def contar_positivos(lista):
    """Retorne quantos números são maiores que zero"""
    contador = 0
    for numero in lista:
        if numero > 0:
            contador = contador + 1
    return contador
```

```python
## Faça seus testes aqui
print(contar_positivos([1, -2, 3, -4, 5]))  # 3
print(contar_positivos([-1, -2, -3]))       # 0
print(contar_positivos([0, 1, 2]))          # 2
```

```python
# Validação
assert contar_positivos([1, -2, 3, -4, 5]) == 3
assert contar_positivos([-1, -2, -3]) == 0
assert contar_positivos([0, 1, 2]) == 2
print("Exercício 2.11 OK!")
```

### 2.15 Calcule a média de uma lista de números

```python
def media(lista_numeros):
    """Retorne a média dos números na lista"""
    soma = 0
    for numero in lista_numeros:
        soma = soma + numero
    return soma / len(lista_numeros)
```

```python
## Faça seus testes aqui
print(media([1, 2, 3, 4, 5]))  # 3.0
print(media([10, 20]))        # 15.0
```

```python
# Validação
assert media([1, 2, 3, 4, 5]) == 3.0
assert media([10, 20]) == 15.0
assert media([5]) == 5.0
print("Exercício 2.15 OK!")
```

### 2.16 Escreva uma função em Python que aceite um inteiro (n) de 1 a 9 e compute o valor de: n+nn+nnn
#### Valide os valores de entrada. Para isso, use a instrução raise ValueError
##### Exemplo: n=5. Então n + nn + nnn = 5 + 55 + 555 = 615

```python
def calculo_n(n):
    """retorne o valor de: n+nn+nnn
    """
    if n < 1 or n > 9:
        raise ValueError("n deve estar entre 1 e 9")
    
    # Converter para string e concatenar
    s = str(n)
    nn = int(s + s)       # Ex: "5" + "5" = "55" -> 55
    nnn = int(s + s + s)  # Ex: "5" + "5" + "5" = "555" -> 555
    
    return n + nn + nnn
```

```python
## Faça seus testes aqui
print(calculo_n(5))  # 615 (5 + 55 + 555)
print(calculo_n(1))  # 123 (1 + 11 + 111)
print(calculo_n(9))  # 1107 (9 + 99 + 999)
```

```python
# Validação para nota
v1 = calculo_n(1) == 123
v2 = calculo_n(2) == 246
v3 = calculo_n(3) == 369
v4 = calculo_n(9) == 1107
print(f"v1={v1}, v2={v2}, v3={v3}, v4={v4}")
print("Exercício 2.16 OK!")
```

### 2.17 Escreva uma função para retornar o maior item de uma lista de inteiros

```python
def maior(lista):
    """retorne o maior item da lista
    """
    maior_valor = lista[0]
    for item in lista:
        if item > maior_valor:
            maior_valor = item
    return maior_valor
```

```python
## Faça seus testes aqui
print(maior([1,0,-1,100,7]))          # 100
print(maior([-100,-1580,-1,-9100,-77]))  # -1
```

```python
# Validação para nota
v1 = maior([1,0,-1,100,7]) == 100
v2 = maior([-100,-1580,-1,-9100,-77]) == -1
print(f"v1={v1}, v2={v2}")
print("Exercício 2.17 OK!")
```

### 2.18 Escreva uma função para contar a quantidade de strings, presentes numa lista, cujo tamanho é maior que 2 e o primeiro e último caracteres são os mesmos

#### Exemplo:  ['abc', 'xyz', 'aba', '1221']
#### Resultado esperado : 2


```python
# Dica 1: A função len retorna o tamanho da string. Exemplo: len('abc'), retorna 3
# Dica 2: Uma string é uma lista de caracteres
def contagem(lista):
    contador = 0
    for texto in lista:
        if len(texto) > 2 and texto[0] == texto[-1]:
            contador = contador + 1
    return contador
```

```python
## Faça seus testes aqui
print(contagem(['abc', 'xyz', 'aba', '1221']))  # 2 (aba e 1221)
```

```python
# Validação para nota
v1 = contagem(['abc', 'xyz', 'aba', '1221']) == 2
v2 = contagem(['ac', '1111', 'aasdfsdfdda', '12ewwww21']) == 3
v3 = contagem(['', 'xyzaaaaaaaaa', 'aa', 'babababa']) == 0
print(f"v1={v1}, v2={v2}, v3={v3}")
print("Exercício 2.18 OK!")
```

### 2.19 Escreva uma função Python para percorrer uma lista e retornar os números multiplos de N. Pare a iteração no ponto em que encontrar um número maior do que K. Não retorne o 1o número maior do que K.
#### Exemplo:  lista = [120, 115, 12, 32, 56, 200, 715, 152, 130, 155, 183, 300], N=5, K=200
#### Resultado Esperado: [120, 115, 200]

```python
def filtra_lista(lista, N, K):
    """retorne uma lista com todos os números multiplos de N menores do que K, 
       até o primeiro número maior do que K, não inclua este número na lista de resultado.
    """
    resultado = []
    for x in lista:
        if x > K:
            break  # Para a iteração quando encontra número maior que K
        if x % N == 0:  # Se x é múltiplo de N
            resultado.append(x)
    return resultado
```

```python
## Faça seus testes aqui
print(filtra_lista([120, 115, 12, 32, 56, 200, 715, 152, 130, 155, 183, 300], 5, 200))
# [120, 115, 200]
```

```python
# Validação para nota
v1 = filtra_lista([120, 115, 12, 32, 56, 200, 715, 152, 130, 155, 183, 300], 5, 200) == [120, 115, 200]
print(f"v1={v1}")
print("Exercício 2.19 OK!")
```
