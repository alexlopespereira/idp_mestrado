## Variáveis

Uma variável é um **rótulo** (nome) associado a um **conteúdo**. O programador escolhe os dois.

```python
myNum = 4
print(myNum)
myNum = 5
print(myNum)
```

## Sintaxe do Python

Os dois-pontos abrem uma **sentença composta**, e o corpo dela é aninhado com espaços ou tabs — não há chaves. O ponto e vírgula no fim da linha é opcional.

> **Erro comum:** esquecer de aninhar o corpo da sentença composta.

```python
x = 2  # Atribuição com sinal de = (igual)
if x > 0:
    # adiciona 1 a x
    x = x + 1;
    print(x)
    print('x maior que zero')
else:
    print('x menor ou igual a zero')
```

## Sintaxe do Python

Funções são declaradas com a palavra-chave `def`. Servem para **reúso** e **legibilidade**. Uma função precisa ser carregada na memória antes de o interpretador conseguir encontrá-la; a chamada se faz com `nome_funcao()`.

```python
def imprimir(a):
    print(a)

imprimir(2)
```

## return de uma função Python

O retorno pode ser **explícito** (`return` seguido de um valor) ou **implícito** — sem `return`, a função devolve `None`.

> **Erro comum:** precisar de um retorno e esquecer de escrevê-lo.

```python
def soma(m, n):
    return m + n
result = soma(4, 5)
print(result)
```

Revisão: primeiro a definição é carregada na memória, depois a chamada executa o corpo.

```python
def soma(a, b):
    s = a + b
    return s
```

```python
result = soma(1, 2)
```

## Laços (loops) do tipo While

O `while` itera enquanto uma condição for verdadeira.

```python
n = 0
total = 0
while n < 4:
    total = total + n*n
    n = n + 1

print(total)
```

## Laços (loops) do tipo for

O `for` itera sobre uma coleção. A palavra-chave `in` referencia a coleção; `val` é um nome escolhido pelo programador. `break` interrompe a iteração.

> Atente para a correta indentação.

```python
sequence = [1, 2, 0, 4, 6, 5, 2, 1]
tot_until_5 = 0
for val in sequence:
    if val == 5:
        break
    tot_until_5 = tot_until_5 + val
print(tot_until_5)
```

## Laço do tipo for com a função range

`range(5)` devolve um iterador da sequência 0, 1, 2, 3, 4. `range(2, 6)` devolve 2, 3, 4, 5.

```python
for val in range(3):
    print(val*val)
```

## Formando uma string (texto)

Strings podem usar aspas simples ou duplas: `'aluno'` e `"aluno"` são equivalentes.

```python
a = 4.5
b = 2
frase1 = f"a={a}, b={b}"
frase2 = "a={0}, b={1}".format(a, b)
print(frase1)
print(frase2)
```

## Funções Populares de String: join

`join` transforma uma lista de strings numa única string, concatenada por um separador.

```python
lista_str = ["1", "2", "3", "4"]
str_concatenado = "-".join(lista_str)
print(str_concatenado)
```

## Funções Populares de String: split

`str.split(sep=None, maxsplit=-1)` divide uma string nas partes separadas por `sep` e devolve uma **lista**.

```python
texto = "1-2-3-4"
lista = texto.split("-")
print(lista)
```

## list (Lista)

`list` é uma sequência de tamanho variável e conteúdo mutável, definida com colchetes `[ ]`. Pode conter objetos de vários tipos. `append` insere; `pop` remove pelo **índice**; `remove` remove pelo **valor**.

```python
al = [2, 4, 0, None]
print(al)
al.append(9)
print(al)
al.pop(1) # Remover pelo índice
print(al)
al.remove(0) # Use remove para remover pelo valor
print(al)
```

## Combinando Listas

O operador `+` concatena duas listas.

```python
al = [2, 4, 0, None]
al_plus = al + ['a', 'b']
print(al_plus)
```

## Slicing (fatiar)

Use intervalos entre colchetes para fatiar sequências.

```python
al = [2, 4, 0, 3, 7, 10, 4, 5]
print(al[0:3])  # Do índice zero até o 3 (não incluso)
print(al[:4])   # Do índice zero ao 4
print(al[2:])   # Do índice 2 até o último
print(al[-1])   # O último elemento
print(al[::2])  # A cada dois elementos a partir do zero
print(al[::-1]) # Reverter/espelhar os elementos
al[1:3] = [8, 8]
print(al)
```

## Tamanho de uma lista: len(lista)

`len` devolve o tamanho da lista.

```python
al = [2, 4, 0, 3, 7, 10, 4, 5]
print(len(al))
```

## Referência: operadores binários

| Operação | Descrição |
|---|---|
| `a + b` | Soma `a` e `b` |
| `a - b` | Subtrai `b` de `a` |
| `a * b` | Multiplica `a` por `b` |
| `a / b` | Divide `a` por `b` |
| `a and b` | `True` se `a` e `b` são `True` |
| `a or b` | `True` se `a` ou `b` é `True` |

| Comparação | Descrição |
|---|---|
| `a == b` | `True` se `a` é igual a `b` |
| `a != b` | `True` se `a` não é igual a `b` |
| `a < b`, `a <= b` | `True` se `a` é menor (ou menor ou igual) que `b` |
| `a > b`, `a >= b` | `True` se `a` é maior (ou maior ou igual) que `b` |
| `a is b` | `True` se `a` e `b` referenciam o mesmo objeto |
| `a is not b` | `True` se `a` e `b` referenciam objetos distintos |

**Tipos escalares:** `None` (valor nulo), `int` (inteiro), `float` (ponto flutuante de
64 bits), `bool` (`True`/`False`). **Tipo sequencial:** `str` (caracteres em UTF-8),
`bytes` (bytes ASCII ou codificados como Unicode).
